```python

import pygame 
import random 
import math 
from pygame import mixer 
import cv2 
import mediapipe as mp 

pygame.init()
window=pygame.display.set_mode((550,800))
pygame.display.set_caption("ROCKET GAME:)")

background=pygame.image.load('background.png').convert()
background=pygame.transform.scale(background,(550,800))

mixer.music.load('background.mp3')
mixer.music.play(-1)

rocket=pygame.image.load("rocket.png")
rocketX=225
rocketY=650
allscores=0
font=pygame.font.Font("freesansbold.ttf",25)
scoreX=20
scoreY=20

cap=cv2.VideoCapture(0)
mp_hands=mp.solutions.hands
hands=mp_hands.Hands(
  max_num_hands=1,
  min_detection_confidence=0.7,
  min_tracking_confidence=0.7
)
mp_draw=mp.solutions.drawing_utils

stone=[]
stoneX=[]
stoneY=[]
stonemoveX=[]
stonemoveY=[]
stoneNumber=5

for i in range(stoneNumber):
  stone.append(pygame.image.load('stone.png'))
  stoneX.append(random.randint(0,500))
  stoneY.append(random.randint(40,250))
  stonemoveX.append(3.5)
  stonemoveY.append(45)

bullet=pygame.image.load('bullet.png')
bulletX=0
bulletY=650
bulletmoveY=18
bullet_state='ready'

def GameOver():
  gameover=font.render("oh...GameOver!!", True , (120,250,45))
  window.blit(gameover,(180,200))

def Score(X,Y):
  score=font.render("Your Score:" + str(allscores), True,(120,250,45))
  window.blit(score,(X,Y))

def Bullet(X,Y):
  global bullet_state
  bullet_state='shoot'
  window.blit(bullet,(X+20,Y+10))

def Rocket (X,Y):
  window.blit(rocket,(X,Y))

def Stone(X,Y,i):
  window.blit(stone[i],(X,Y))

def Fire(stoneX,stoneY,bulletX,bulletY):
  d=math.sqrt((math.pow(stoneX-bulletX,2))+(math.pow(stoneY-bulletY,2)))
  if d<20:
    return True
  else:
    return False

def get_hand_data():
  ret,frame=cap.read()
  if not ret:
    return None , False
  frame=cv2.flip(frame,1)
  rgb=cv2.cvtColor(frame,cv2.COLOR_BGR2RGB)
  results=hands.process(rgb)
  if results.multi_hand_landmarks:
    hand=results.multi_hand_landmarks[0]
    wrist=hand.landmark[0]
    hand_x=wrist.x
    rocket_position=int(hand_x *550)
    fingers=[(8,6),
             (12,10),
             (16,14),
             (20,18)]
    closed_fingers=0
    for tip,pip in fingers:
      if (hand.landmark[tip].y > hand.landmark[pip].y):
        closed_fingers+=1
    fist=closed_fingers>=3
    mp_draw.draw_landmarks(frame,hand,mp_hands.HAND_CONNECTIONS)
    cv2.imshow("Hand Control",frame)
    return rocket_position ,fist
  cv2.imshow("Hand Control",frame)
  return None , False

r=True
while r:
  window.blit(background,(0,0))
  for event in pygame.event.get():
    if event.type==pygame.QUIT:
      r=False
  hand_x,fist=get_hand_data()
  if hand_x is not None:
    rocketX=hand_x - 25
  if rocketX<0:
    rocketX=0
  elif rocketX>470:
    rocketX=470
  if fist and bullet_state=="ready":
    bulletX=rocketX
    bulletY=rocketY
    bullet_state="shoot"
    bulletsound=mixer.Sound("pew.mp3")
    bulletsound.play()
  for i in range (stoneNumber):
    if stoneY[i]>600:
      for j in range(stoneNumber):
        stoneY[j]=5000
      GameOver()
      break
    stoneX[i]=stoneX[i]+stonemoveX[i]
    if stoneX[i]<=0:
      stonemoveX[i]=3.5
      stoneY[i]=stoneY[i]+stonemoveY[i]
    
    elif stoneX[i]>=490:
      stonemoveX[i]=-3.5
      stoneY[i]=stoneY[i]+stonemoveY[i]
    
    fire=Fire(stoneX[i],stoneY[i],bulletX,bulletY)
    if fire:
      bulletY=480
      bullet_state='ready'
      allscores+=1
      stoneX[i]=random.randint(0,500)
      stoneY[i]=random.randint(40,250)
      firesound=mixer.Sound('fire.mp3')
      firesound.play()
    Stone(stoneX[i],stoneY[i],i)
  
  if bulletY<=0:
    bulletY=480
    bullet_state='ready'
  if bullet_state is 'shoot':
    Bullet(bulletX,bulletY)
    bulletY-=bulletmoveY

  
  Score(scoreX,scoreY)
  Rocket(rocketX,rocketY)
  pygame.display.update()

cap.release()
cv2.destroyAllWindows()
pygame.quit()











```
